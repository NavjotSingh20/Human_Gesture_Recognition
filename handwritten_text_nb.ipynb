```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.metrics import accuracy_score, confusion_matrix
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.datasets import fetch_openml
```

```python
mnist = fetch_openml('mnist_784', version=1, as_frame=False)

X = mnist.data
y = mnist.target.astype(int)

X = X / 255.0

X_train, X_test = X[:60000], X[60000:]
y_train, y_test = y[:60000], y[60000:]
```

```python
lr1 = LogisticRegression(C=1.0, max_iter=250)
lr2 = LogisticRegression(C=0.1, max_iter=300)

lr1.fit(X_train, y_train)
lr2.fit(X_train, y_train)

pred_lr1 = lr1.predict(X_test)
pred_lr2 = lr2.predict(X_test)

print("LR C=1.0 Accuracy:", accuracy_score(y_test, pred_lr1))
print("LR C=0.1 Accuracy:", accuracy_score(y_test, pred_lr2))
```

```python
plt.figure(figsize=(6,4))
sns.heatmap(confusion_matrix(y_test, pred_lr1), fmt='d', cmap='Blues')
plt.title("Logistic Regression Confusion Matrix (C = 1.0)")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()

plt.figure(figsize=(6,4))
sns.heatmap(confusion_matrix(y_test, pred_lr2), fmt='d', cmap='Blues')
plt.title("Logistic Regression Confusion Matrix (C = 0.1)")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

```python
knn3 = KNeighborsClassifier(n_neighbors=3)
knn7 = KNeighborsClassifier(n_neighbors=7)

knn3.fit(X_train, y_train)
knn7.fit(X_train, y_train)

pred_knn3 = knn3.predict(X_test)
pred_knn7 = knn7.predict(X_test)

print("KNN k=3 Accuracy:", accuracy_score(y_test, pred_knn3))
print("KNN k=7 Accuracy:", accuracy_score(y_test, pred_knn7))
```

```python
plt.figure(figsize=(6,4))
sns.heatmap(confusion_matrix(y_test, pred_knn3), fmt='d', cmap='Greens')
plt.title("KNN Confusion Matrix (k = 3)")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()

plt.figure(figsize=(6,4))
sns.heatmap(confusion_matrix(y_test, pred_knn7), fmt='d', cmap='Greens')
plt.title("KNN Confusion Matrix (k = 7)")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

```python
dt10 = DecisionTreeClassifier(max_depth=10)
dt20 = DecisionTreeClassifier(max_depth=20)

dt10.fit(X_train, y_train)
dt20.fit(X_train, y_train)

pred_dt10 = dt10.predict(X_test)
pred_dt20 = dt20.predict(X_test)

print("DT depth=10 Accuracy:", accuracy_score(y_test, pred_dt10))
print("DT depth=20 Accuracy:", accuracy_score(y_test, pred_dt20))
```

```python
plt.figure(figsize=(6,4))
sns.heatmap(confusion_matrix(y_test, pred_dt10), fmt='d', cmap='Oranges')
plt.title("Decision Tree Confusion Matrix (max_depth = 10)")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()

plt.figure(figsize=(6,4))
sns.heatmap(confusion_matrix(y_test, pred_dt20), fmt='d', cmap='Oranges')
plt.title("Decision Tree Confusion Matrix (max_depth = 20)")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```
